```python
# -*- coding: utf-8 -*-
"""Buoc 1: thay accuracy danh lua tren du lieu lech lop."""
import pandas as pd
from sklearn.dummy import DummyClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
from sklearn.model_selection import train_test_split
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler

df = pd.read_csv("/content/sinh_vien_bo_hoc.csv")
print("Kich thuoc bang:", df.shape)
print(df.head())
print()
print("So sinh vien moi lop:")
print(df["bo_hoc"].value_counts().to_string())
print(f"Ty le bo hoc: {df['bo_hoc'].mean():.3f}")
print()

X = df.drop(columns="bo_hoc")
y = df["bo_hoc"]
X_tr, X_te, y_tr, y_te = train_test_split(
    X, y, test_size=0.25, random_state=17, stratify=y
)
print(f"Tap kiem tra: {len(y_te)} ban, {y_te.sum()} ban bo hoc")
print()

luoi = DummyClassifier(strategy="most_frequent").fit(X_tr, y_tr)
du_luoi = luoi.predict(X_te)

lr = make_pipeline(StandardScaler(), LogisticRegression(max_iter=1000))
lr.fit(X_tr, y_tr)
du_lr = lr.predict(X_te)

for ten, du in [("Mo hinh luoi", du_luoi), ("Hoi quy logistic", du_lr)]:
    acc = accuracy_score(y_te, du)
    tim_ra = ((du == 1) & (y_te == 1)).sum()
    print(f"{ten:17s} accuracy {acc:.4f}, "
          f"tim ra {tim_ra} tren {y_te.sum()} ban bo hoc")
```

```python
# -*- coding: utf-8 -*-
"""Buoc 2: tu dem bon o cua ma tran nham lan, roi so voi thu vien."""
import numpy as np
import pandas as pd
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import confusion_matrix
from sklearn.model_selection import train_test_split
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler

df = pd.read_csv("sinh_vien_bo_hoc.csv")
X = df.drop(columns="bo_hoc")
y = df["bo_hoc"]
X_tr, X_te, y_tr, y_te = train_test_split(
    X, y, test_size=0.25, random_state=17, stratify=y
)

lr = make_pipeline(StandardScaler(), LogisticRegression(max_iter=1000))
du = lr.fit(X_tr, y_tr).predict(X_te)
that = y_te.to_numpy()

tp = int(np.sum((that == 1) & (du == 1)))
fp = int(np.sum((that == 0) & (du == 1)))
fn = int(np.sum((that == 1) & (du == 0)))
tn = int(np.sum((that == 0) & (du == 0)))
print("Tu dem:")
print(f" TP = {tp} (doan bo hoc, dung la bo hoc)")
print(f" FP = {fp} (bao dong nham)")
print(f" FN = {fn} (bo sot)")
print(f" TN = {tn} (doan khong bo, dung la khong bo)")
print(f" Tong = {tp + fp + fn + tn}")
print()

m = confusion_matrix(that, du)
print("confusion_matrix cua scikit-learn:")
print(m)
print("Thu tu khi trai ra: TN, FP, FN, TP =", m.ravel())
print()

acc = (tp + tn) / (tp + fp + fn + tn)
print(f"Accuracy tinh tu bon o: ({tp} + {tn}) / {len(that)} = {acc:.4f}")
print(f"So ban bi bao co nguy co: TP + FP = {tp + fp}")
```

```python
# -*- coding: utf-8 -*-
"""Buoc 3: precision, recall, F1 tinh tay va tinh bang thu vien."""
import pandas as pd
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import (classification_report, confusion_matrix,
                             f1_score, precision_score, recall_score)
from sklearn.model_selection import train_test_split
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler

df = pd.read_csv("sinh_vien_bo_hoc.csv")
X = df.drop(columns="bo_hoc")
y = df["bo_hoc"]
X_tr, X_te, y_tr, y_te = train_test_split(
    X, y, test_size=0.25, random_state=17, stratify=y
)

lr = make_pipeline(StandardScaler(), LogisticRegression(max_iter=1000))
du = lr.fit(X_tr, y_tr).predict(X_te)

tn, fp, fn, tp = confusion_matrix(y_te, du).ravel()
p = tp / (tp + fp)
r = tp / (tp + fn)
f1 = 2 * p * r / (p + r)
print("Tinh tay tu bon o:")
print(f" Precision = {tp} / ({tp} + {fp}) = {p:.4f}")
print(f" Recall    = {tp} / ({tp} + {fn}) = {r:.4f}")
print(f" F1 = 2 x {p:.4f} x {r:.4f} / ({p:.4f} + {r:.4f}) = {f1:.4f}")
print()

print("Tinh bang scikit-learn:")
print(f" Precision = {precision_score(y_te, du):.4f}")
print(f" Recall    = {recall_score(y_te, du):.4f}")
print(f" F1        = {f1_score(y_te, du):.4f}")
print()

print("Bao cao day du:")
print(classification_report(y_te, du, digits=4,
                            target_names=["khong bo", "bo hoc"]))
```

```python
# -*- coding: utf-8 -*-
"""Buoc 4: doi nguong, va chon nguong dung cach tren tap hoc."""
import numpy as np
import pandas as pd
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import f1_score, precision_score, recall_score
from sklearn.model_selection import cross_val_predict, train_test_split
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler

df = pd.read_csv("sinh_vien_bo_hoc.csv")
X = df.drop(columns="bo_hoc")
y = df["bo_hoc"]
X_tr, X_te, y_tr, y_te = train_test_split(
    X, y, test_size=0.25, random_state=17, stratify=y
)

lr = make_pipeline(StandardScaler(), LogisticRegression(max_iter=1000))
lr.fit(X_tr, y_tr)
xs = lr.predict_proba(X_te)[:, 1]

print("Tap kiem tra, doi nguong")
print("nguong  so bao  TP  FP  FN   precision  recall    F1")
for t in [0.5, 0.4, 0.3, 0.2, 0.1]:
    du = (xs >= t).astype(int)
    tp = int(((du == 1) & (y_te == 1)).sum())
    fp = int(((du == 1) & (y_te == 0)).sum())
    fn = int(((du == 0) & (y_te == 1)).sum())
    print(f" {t:.1f}    {du.sum():4d}  {tp:3d} {fp:3d} {fn:3d}   "
          f"{precision_score(y_te, du):.4f}    "
          f"{recall_score(y_te, du):.4f}    "
          f"{f1_score(y_te, du):.4f}")
print()

# Chon nguong dung cach: chi dung tap hoc, qua kiem dinh cheo
xs_hoc = cross_val_predict(lr, X_tr, y_tr, cv=5,
                           method="predict_proba")[:, 1]
cac_nguong = np.round(np.arange(0.05, 0.96, 0.05), 2)
f1_hoc = [f1_score(y_tr, (xs_hoc >= t).astype(int)) for t in cac_nguong]
t_tot = cac_nguong[int(np.argmax(f1_hoc))]
print(f"Nguong cho F1 cao nhat tren tap hoc: {t_tot} "
      f"(F1 kiem dinh cheo {max(f1_hoc):.4f})")

# Chon xong moi dung tap kiem tra, va chi mot lan
du = (xs >= t_tot).astype(int)
print(f"Tap kiem tra voi nguong {t_tot}: "
      f"precision {precision_score(y_te, du):.4f}, "
      f"recall {recall_score(y_te, du):.4f}, "
      f"F1 {f1_score(y_te, du):.4f}")
```

```python
# -*- coding: utf-8 -*-
"""Buoc 5: duong ROC, AUC, va so cac mo hinh da hoc trong mon."""
import pandas as pd
import matplotlib
matplotlib.use("Agg")          # chay khong can man hinh
import matplotlib.pyplot as plt

from sklearn.ensemble import RandomForestClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import (accuracy_score, f1_score, precision_score,
                             recall_score, roc_auc_score, roc_curve)
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import GaussianNB
from sklearn.neighbors import KNeighborsClassifier
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.svm import SVC
from sklearn.tree import DecisionTreeClassifier

df = pd.read_csv("sinh_vien_bo_hoc.csv")
X = df.drop(columns="bo_hoc")
y = df["bo_hoc"]
X_tr, X_te, y_tr, y_te = train_test_split(
    X, y, test_size=0.25, random_state=17, stratify=y
)

cac_mo_hinh = {
    "Logistic":        make_pipeline(StandardScaler(),
                                     LogisticRegression(max_iter=1000)),
    "KNN (k=7)":       make_pipeline(StandardScaler(),
                                     KNeighborsClassifier(n_neighbors=7)),
    "Naive Bayes":     make_pipeline(StandardScaler(), GaussianNB()),
    "Cay quyet dinh":  DecisionTreeClassifier(max_depth=5, random_state=0),
    "Rung ngau nhien": RandomForestClassifier(n_estimators=200,
                                              random_state=0, n_jobs=-1),
    "SVM (RBF)":       make_pipeline(StandardScaler(),
                                     SVC(probability=True, random_state=0)),
}

plt.figure(figsize=(7, 6))
print(f"{'Mo hinh':18s} {'AUC':>8s} {'Recall':>8s} {'F1':>8s}")
print("-" * 46)
for ten, mh in cac_mo_hinh.items():
    mh.fit(X_tr, y_tr)
    xs = mh.predict_proba(X_te)[:, 1]
    fpr, tpr, _ = roc_curve(y_te, xs)
    auc = roc_auc_score(y_te, xs)
    plt.plot(fpr, tpr, label=f"{ten} (AUC = {auc:.3f})")

    du = (xs >= 0.5).astype(int)
    print(f"{ten:18s} {auc:8.4f} "
          f"{recall_score(y_te, du):8.4f} "
          f"{f1_score(y_te, du):8.4f}")

plt.plot([0, 1], [0, 1], "k--", label="Doan ngau nhien (AUC = 0.5)")
plt.xlabel("Ty le bao dong nham (FPR)")
plt.ylabel("Recall (TPR)")
plt.title("Duong ROC tren tap kiem tra")
plt.legend(loc="lower right")
plt.grid(alpha=0.3)
plt.tight_layout()
plt.savefig("roc_so_sanh.png", dpi=120)
print("\nDa luu hinh: roc_so_sanh.png")
```

```python
# -*- coding: utf-8 -*-
"""Buoc 6: kiem dinh cheo nhieu thuoc do va can bang lop khi hoc."""
import pandas as pd
from sklearn.dummy import DummyClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report
from sklearn.model_selection import cross_val_score, train_test_split
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler

df = pd.read_csv("sinh_vien_bo_hoc.csv")
X = df.drop(columns="bo_hoc")
y = df["bo_hoc"]

# 7.1 Vi sao phai chia co phan tang
print("=== 7.1 So sanh chia thuong va chia phan tang ===")
for ten, strat in [("Khong stratify", None), ("Co stratify", y)]:
    X_tr, X_te, y_tr, y_te = train_test_split(
        X, y, test_size=0.25, random_state=17, stratify=strat
    )
    print(f"{ten:16s}: tap kiem tra {y_te.sum():3d}/{len(y_te)} ban bo hoc "
          f"(ty le {y_te.mean():.3f})")
print()

# 7.2 Kiem dinh cheo voi nhieu thuoc do
print("=== 7.2 Kiem dinh cheo 5-fold voi nhieu thuoc do ===")
X_tr, X_te, y_tr, y_te = train_test_split(
    X, y, test_size=0.25, random_state=17, stratify=y
)

luoi = DummyClassifier(strategy="most_frequent")
lr_thuong = make_pipeline(StandardScaler(), LogisticRegression(max_iter=1000))

for ten, mh in [("Mo hinh luoi", luoi), ("Logistic thuong", lr_thuong)]:
    acc = cross_val_score(mh, X_tr, y_tr, cv=5, scoring="accuracy").mean()
    p   = cross_val_score(mh, X_tr, y_tr, cv=5, scoring="precision").mean()
    r   = cross_val_score(mh, X_tr, y_tr, cv=5, scoring="recall").mean()
    f1  = cross_val_score(mh, X_tr, y_tr, cv=5, scoring="f1").mean()
    print(f"{ten:16s} acc {acc:.4f}  precision {p:.4f}  "
          f"recall {r:.4f}  F1 {f1:.4f}")
print()

# 7.3 Can bang lop khi hoc
print("=== 7.3 Can bang lop khi hoc: class_weight='balanced' ===")
lr_can_bang = make_pipeline(
    StandardScaler(),
    LogisticRegression(max_iter=1000, class_weight="balanced")
)

for ten, mh in [("Logistic thuong", lr_thuong),
                ("Logistic can bang", lr_can_bang)]:
    mh.fit(X_tr, y_tr)
    du = mh.predict(X_te)
    print(f"--- {ten} ---")
    print(classification_report(y_te, du, digits=4,
                                target_names=["khong bo", "bo hoc"]))
```
