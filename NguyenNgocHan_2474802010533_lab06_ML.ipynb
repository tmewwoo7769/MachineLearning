Bai 01: Khong cach toi ranh gioi

```python
# -*- coding: utf-8 -*-
"""Bai tap 1: Tinh tay khoang cach toi ranh gioi."""
import numpy as np

w1, w2, b = -0.831, -0.355, 4.820
do_dai_w = np.sqrt(w1**2 + w2**2)
le = 2 / do_dai_w

def tinh_f_va_khoang_cach(x1, x2):
    f = w1*x1 + w2*x2 + b
    kc = abs(f) / do_dai_w
    return f, kc

print(f"Do rong le = {le:.4f}")
print("Diem (3, 3):")
f1, kc1 = tinh_f_va_khoang_cach(3, 3)
print(f"  f(x) = {f1:.3f}, khoang cach = {kc1:.4f}")

print("Diem (5, 4.5):")
f2, kc2 = tinh_f_va_khoang_cach(5, 4.5)
print(f"  f(x) = {f2:.3f}, khoang cach = {kc2:.4f}")

print("Ba vecto ho tro (co f(x) = 1 hoac -1):")
print(f"  Khoang cach = 1 / {do_dai_w:.4f} = {1/do_dai_w:.4f}")
```

Bai 02: Them diem

```python
# -*- coding: utf-8 -*-
"""Bai tap 2: Them mot diem vao du lieu."""
import numpy as np
from sklearn.datasets import make_blobs
from sklearn.svm import SVC

X, y = make_blobs(n_samples=24, centers=[[2, 2], [6, 6]],
                  cluster_std=0.9, random_state=0)
y = 1 - y

# Truong hop 1: Them diem (-1, -1), lop 1
X_moi_1 = np.vstack([X, [-1, -1]])
y_moi_1 = np.append(y, 1)
svm_1 = SVC(kernel="linear", C=1000).fit(X_moi_1, y_moi_1)
print("Them diem (-1, -1):")
print(f"  w1 = {svm_1.coef_[0][0]:.3f}, w2 = {svm_1.coef_[0][1]:.3f}, b = {svm_1.intercept_[0]:.3f}")

# Truong hop 2: Bo diem tren, them diem (4.5, 4), lop 1
X_moi_2 = np.vstack([X, [4.5, 4]])
y_moi_2 = np.append(y, 1)
svm_2 = SVC(kernel="linear", C=1000).fit(X_moi_2, y_moi_2)
print("Them diem (4.5, 4):")
print(f"  w1 = {svm_2.coef_[0][0]:.3f}, w2 = {svm_2.coef_[0][1]:.3f}, b = {svm_2.intercept_[0]:.3f}")
```

bai 03: kernel da thuc

```python
# -*- coding: utf-8 -*-
"""Bai tap 3: Kernel da thuc."""
import pandas as pd
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.svm import SVC

df = pd.read_csv("hai_vang_trang.csv")
X = df[["x1", "x2"]]
y = df["nhan"]
X_tr, X_te, y_tr, y_te = train_test_split(
    X, y, test_size=0.25, random_state=17, stratify=y
)

for d in [2, 3, 4]:
    print(f"--- Kernel da thuc bac {d} ---")
    pipe = make_pipeline(StandardScaler(), SVC(kernel="poly", degree=d))
    luoi = {"svc__C": [0.1, 1, 10, 100]} # Luu y ten tham so
    tim = GridSearchCV(pipe, luoi, cv=5)
    tim.fit(X_tr, y_tr)
    print(f"  Cap tot nhat: {tim.best_params_}")
    print(f"  Diem CV tot nhat: {tim.best_score_:.4f}")
    print(f"  Accuracy tap kiem tra: {tim.score(X_te, y_te):.4f}")
```

Bai 04: dò lưới

```python
# -*- coding: utf-8 -*-
"""Bai tap 4: Do lui min."""
import pandas as pd
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.svm import SVC

df = pd.read_csv("hai_vang_trang.csv")
X = df[["x1", "x2"]]
y = df["nhan"]
X_tr, X_te, y_tr, y_te = train_test_split(
    X, y, test_size=0.25, random_state=17, stratify=y
)

luoi_min = {"C": [3, 10, 30], "gamma": [0.3, 1, 3]}
tim = GridSearchCV(SVC(kernel="rbf"), luoi_min, cv=5)
tim.fit(X_tr, y_tr)

print("Cap tot nhat sau khi do min:", tim.best_params_)
print(f"Diem CV tot nhat: {tim.best_score_:.4f}")
print(f"Accuracy tap kiem tra: {tim.score(X_te, y_te):.4f}")
```

bai 05: viet tay, lỗi

```python
# -*- coding: utf-8 -*-
"""Bai tap 5: Bang nham lan chu so."""
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import load_digits
from sklearn.model_selection import train_test_split
from sklearn.svm import SVC
from sklearn.metrics import confusion_matrix

X, y = load_digits(return_X_y=True)
X_tr, X_te, y_tr, y_te = train_test_split(
    X, y, test_size=0.25, random_state=17, stratify=y
)

svm = SVC(kernel="rbf").fit(X_tr, y_tr)
du_doan = svm.predict(X_te)

# In bang nham lan
cm = confusion_matrix(y_te, du_doan)
print("Bang nham lan 10x10:")
print(cm)

# Tim va luu anh bi doan sai
sai = np.where(du_doan != y_te)[0]
print("\nCac cap nham lan:")
for i in sai:
    print(f"  Anh {i}: that la {y_te[i]}, doan la {du_doan[i]}")
    plt.figure()
    plt.imshow(X_te[i].reshape(8, 8), cmap="gray_r")
    plt.title(f"That: {y_te[i]} - Doan: {du_doan[i]}")
    plt.savefig(f"loi_anh_{i}.png") # Luu anh de xem
    plt.close()
```

bai 06: SVM tren du lieu lon

```python
# -*- coding: utf-8 -*-
"""Bai tap 6: Thoi gian hoc cua SVC va LinearSVC."""
import time
import pandas as pd
from sklearn.datasets import make_classification
from sklearn.svm import SVC, LinearSVC

mang_n = [2000, 5000, 10000, 20000]
ket_qua = []

for n in mang_n:
    X, y = make_classification(n_samples=n, n_features=20, random_state=0)

    # Do thoi gian SVC
    svc = SVC()
    t1 = time.time()
    svc.fit(X, y)
    t_svc = time.time() - t1

    # Do thoi gian LinearSVC
    lsvc = LinearSVC()
    t2 = time.time()
    lsvc.fit(X, y)
    t_lsvc = time.time() - t2

    ket_qua.append({"n": n, "SVC (s)": round(t_svc, 3), "LinearSVC (s)": round(t_lsvc, 3)})

df_kq = pd.DataFrame(ket_qua)
print(df_kq.to_string(index=False))
```
